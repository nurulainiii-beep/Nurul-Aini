```python
volume = [35,42,48,45]
total = sum(volume)
rata_rata =  sum(volume)/len(volume)
tertinggi = max(volume)
terendah = min(volume)

print("sum:", total)
print("average:", rata_rata)
print("higher:", tertinggi)
print("lower:", terendah)
```

```python
import numpy as anjay

volume = anjay.array([35,42,48,45])
print("sum:", anjay.sum(volume))
print("average:", anjay.average(volume))
print("max:", anjay.max(volume))
print("min:", anjay.min(volume))
```

```python
import numpy as cihuy

dowo = cihuy.array([4.0, 5.0, 6.0])  #meter
wero =  cihuy.array([3.0, 3.0, 2.5]) #meter
tebal = 0.12                         #meter

luas = dowo*wero
volume = luas*tebal

print("luas tiap pelat:", luas, "m2")
print("volume tiap pelat:", volume, "m3")
print("total volume:", cihuy.sum(volume), "m3")
```
